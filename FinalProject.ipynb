## Abstract

This study investigates how macroeconomic conditions and media sentiment together predict the scale of tech-sector layoffs, with a particular focus on whether AI-focused companies respond differently from traditional tech firms. Between 2020 and 2026, the tech industry experienced dramatic swings in employment, including a COVID-era hiring boom followed by a wave of mass layoffs starting in 2022. Pursuing our curiosity about AI's role in reshaping the labor market, we build on Shekhar and Saurombe's (2026) work on "algorithmic anxiety" by quantifying surface-level media sentiment via VADER scores and testing whether AI companies are more sensitive to shifts in public perception than to macroeconomic signals, compared to traditional tech.


To investigate this, we merged three datasets: a layoff events dataset (1,590 cleaned events from 2020–2026), VADER-based news sentiment scores from 306 articles, and monthly Federal Funds Rate data from the Federal Reserve. After cleaning for missingness and joining on date-based keys, we conducted exploratory data analysis and a preliminary multiple regression predicting log-transformed layoff count using FFR, sentiment score, and an AI-sector indicator on a merged sample of 73 events.


Our EDA revealed that layoff activity peaked sharply in early 2023 for both sectors simultaneously, suggesting macroeconomic shocks affect AI and traditional tech companies at the same time rather than independently. News sentiment averaged slightly negative (mean VADER score: -0.047), consistent with underlying anxiety about job displacement. Our preliminary regression on the merged sample (n=73) found that FFR and sentiment explain limited variance in layoff scale, and the AI-sector interaction term was not strongly significant, though the small merged sample constrains our conclusions.

## Research Question

How do macroeconomic indicators (such as the Federal Funds Rate and CPI) and news sentiment scores interact to predict the scale of layoffs in AI-focused companies compared to traditional tech companies?
Specifically, we will use a multiple regression analysis to determine if AI-sector layoffs are more sensitive to shifts in public sentiment (VADER sentiment scores) or to economic signals (Federal Funds Rate). Our analysis will use the total number of employees laid off per event as the scale and identify "AI-focused" companies based on the industry classification provided in the dataset.

## Background and Prior Work

Between 2019 and 2021, there was a massive wave of employment in the tech sector. Companies were doubling their headcount and going through rapid hiring to meet consumer demand that peaked during the COVID era. However in 2022 it all changed; there were mass layoffs at large companies including Google, Meta, and Amazon. In fact researchers found that contrary to popular belief, the reason for these layoffs wasn’t AI, but macroeconomic factors, specifically the Federal Reserve's monetary tightening and increase of interest rates (which peaked in 2022) to combat inflation <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1). When interest rates rose, labor became expensive while revenue decreased causing layoffs. In more general tech companies this meant just massive layoffs due to the inability to make “easy money” and to stay profitable or completely shutting down their company <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). However, for large AI-based companies like Meta, over 20% of the workforce was cut to pay for servers that run AI <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1).

Digital and news sentiment has also played a large role in the labor market, especially in tech companies. In their research, Shekhar and Saurombe (2026) identify a phenomenon called "algorithmic anxiety," where employees perceive AI as a breach of the traditional psychological contract between employers and employees. This leads to lower productivity among the workers which eventually leads to employers replacing them with AI. Their study goes further into the measure of sentiment by comparing that while basic lexicon tools (VADER) often show positive surface-level optimism about AI, transformer-based models (BERT) uncover more deep-seated negative sentiment (51%) related to job displacement and “corporate betrayal” as companies invest more in AI <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). In our project, we will build upon Shekhar and Saurombe’s (2026) findings by using VADER to quantify the 'surface-level' sentiment in our news dataset and comparing it against the actual layoff numbers.

Along with macroeconomic factors, this underlying sentiment may act as a predictive catalyst. This threat is also higher in AI-focused companies, however, most research does not provide much detail into how the layoff statistics are different in AI-based and general tech companies. This distinction is important because AI-based companies layoff employees to fund AI infrastructure, but general tech companies layoff to stay profitable so the types of jobs lost might be different. There is also differing perspective as the research sites that media sentiment may be biased because those who have been affected by the layoffs are more likely to share their opinions <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3). Furthermore, there is also significant research that AI is not responsible for the layoffs, even in tech companies. Researchers claim that AI isn’t equipped or trained enough to take over most jobs, but does have the potential to take over the entry-level jobs <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4). Essentially by clarifying whether layoffs are driven by tangible economic shifts (rates) or psychological hype (sentiment), our findings could help job seekers and policymakers better predict industry stability during technological pivots.

## Hypothesis


We predict that layoffs in AI-focused companies will show a stronger correlation with news sentiment scores than with the Federal Funds Rate, whereas traditional tech companies will be more sensitive to macroeconomic signals.

While traditional tech firms prioritize profitability and debt management in high-interest environments, AI-focused companies are more driven by technological change where employment levels are heavily influenced by the public perception of innovation. ### Dataset #1: AI & Tech Layoffs Events (2020–Present)

This dataset records individual layoff events across AI-focused and traditional tech companies from 2020 to the present. Each row represents a single layoff event at a specific company on a specific date, yielding 2,470 observations across 11 variables.

The primary dependent variable is `layoff_count`, which measures the total number of employees laid off in a single event as a raw headcount (e.g., a value of 11,000 means 11,000 individual employees were let go). Notably, approximately 35% of `layoff_count` values are missing, likely because some companies disclosed layoffs without releasing exact headcount figures. A related variable, `pct_workforce`, captures the share of the total workforce affected (e.g., `"10%"`), but it is stored as a string with a `%` suffix and has even higher missingness (~38%). We will use `layoff_count` as our primary outcome variable and convert `pct_workforce` to numeric as a supplementary measure where available.

The key grouping variable is `is_ai_company`, a boolean that classifies each company as AI-focused (`True`) or traditional tech (`False`). This variable has no missing values and is the basis for our sector comparison. The `stage` variable records the company's funding stage at the time of the layoff (e.g., Seed, Series B, Post-IPO), which serves as a potential control variable for company maturity and financial position. `raised_mm` captures total funding raised in millions of dollars, but is stored as a string (e.g., `"$45.00"`) and must be converted to a numeric type before use.

The `date` variable records when the layoff occurred in YYYY-MM-DD format and is essential for joining this dataset with the news sentiment table. The `industry` and `country` columns provide context for stratified analysis and have only 2 missing values each. The `source_url` column links to the original report of each layoff event and is not used directly in modeling.

The main concerns with this dataset are the high missingness in `layoff_count` (~35%) and `pct_workforce` (~38%), the string formatting of `raised_mm`, and the binary nature of `is_ai_company`, which may not fully capture companies operating across both AI and traditional business lines (e.g., Google, Meta). Additionally, because data is sourced from publicly reported events, smaller or private companies may be underrepresented.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# A. Load raw data
df = pd.read_csv('data/00-raw/layoffs_events.csv')

# B & C. Tidy check + Size
# Tidy data: each row = one layoff event, each column = one variable.
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"\nColumn dtypes:\n{df.dtypes}")
df.head()
```

```python
# D. Missingness analysis
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(1)
missing_df = pd.DataFrame({'missing_count': missing, 'missing_%': missing_pct})
missing_df = missing_df[missing_df['missing_count'] > 0].sort_values('missing_%', ascending=False)
print("Columns with missing values:")
print(missing_df)

# Visualize missingness pattern
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart of missing %
axes[0].barh(missing_df.index, missing_df['missing_%'], color='steelblue')
axes[0].set_xlabel('Missing (%)')
axes[0].set_title('Missing Data by Column — layoffs_events')
for i, v in enumerate(missing_df['missing_%']):
    axes[0].text(v + 0.3, i, f'{v}%', va='center')

# Is missingness in layoff_count related to is_ai_company?
miss_by_ai = df.groupby('is_ai_company')['layoff_count'].apply(lambda x: x.isnull().mean() * 100).round(1)
axes[1].bar(miss_by_ai.index.map({True: 'AI Company', False: 'Traditional Tech'}),
            miss_by_ai.values, color=['#e07b54', '#5b8db8'])
axes[1].set_ylabel('% Missing layoff_count')
axes[1].set_title('layoff_count Missingness by Company Type')
for i, v in enumerate(miss_by_ai.values):
    axes[1].text(i, v + 0.5, f'{v}%', ha='center')

plt.tight_layout()
plt.show()

print("\nMissingness rates for layoff_count by is_ai_company:")
print(miss_by_ai)
```

```python
# E. Outlier detection
lc = df['layoff_count'].dropna()
q1, q3 = lc.quantile(0.25), lc.quantile(0.75)
upper_fence = q3 + 3 * (q3 - q1)

print(f"layoff_count summary:\n{lc.describe().round(0)}\n")
print(f"Extreme outlier threshold (3×IQR): {upper_fence:.0f}")
print(f"\nEvents above threshold:")
print(df[df['layoff_count'] > upper_fence][['company', 'date', 'layoff_count']]
      .sort_values('layoff_count', ascending=False).to_string(index=False))

# Boxplot
plt.figure(figsize=(6, 4))
plt.boxplot(lc, vert=False)
plt.xlabel('layoff_count')
plt.title('layoff_count Distribution (with outliers)')
plt.tight_layout()
plt.show()

# These are real, well-documented events (e.g. Amazon, Meta) — not data errors.
# We retain them but will probably log-transform layoff_count in modeling to handle skew.
```

```python
# F. Clean & tidy
df_clean = df.copy()

# 1. Parse date -> datetime
df_clean['date'] = pd.to_datetime(df_clean['date'])

# 2. Parse raised_mm: strip '$', convert to float
df_clean['raised_mm'] = (
    df_clean['raised_mm'].astype(str)
    .str.replace('$', '', regex=False).str.replace(',', '', regex=False)
    .str.strip().replace('nan', np.nan).astype(float)
)

# 3. Parse pct_workforce: strip '%', convert to float
df_clean['pct_workforce'] = (
    df_clean['pct_workforce'].astype(str)
    .str.replace('%', '', regex=False).str.strip()
    .replace('nan', np.nan).astype(float)
)

# 4. Drop rows missing layoff_count (our DV — cannot impute outcome variable)
n_before = len(df_clean)
df_clean = df_clean.dropna(subset=['layoff_count'])
print(f"Dropped {n_before - len(df_clean)} rows with missing layoff_count")

# 5. Drop the 2 rows missing industry/country
df_clean = df_clean.dropna(subset=['industry', 'country'])

# 6. Fill 5 missing stage values with 'Unknown'
df_clean['stage'] = df_clean['stage'].fillna('Unknown')

key_cols = ['company', 'date', 'layoff_count', 'is_ai_company', 'industry', 'country', 'stage']
print(f"\nRemaining nulls in key columns:\n{df_clean[key_cols].isnull().sum()}")
print(f"\nFinal shape: {df_clean.shape[0]} rows × {df_clean.shape[1]} columns")
df_clean.head()
```

### Dataset #2: Media Sentiment Analysis of Global Layoff News

This dataset contains 306 observations and 10 variables related to media coverage of AI-related layoffs and labor market trends from 2020 onward. The dataset focuses on sentiment analysis of news articles related to AI layoffs and the technology job market from 2020 onward. Each row represents a single news article with variables including publication date, sentiment score, sentiment category, and whether the article specifically discussed layoffs.

The primary variable is sentiment, a VADER compound sentiment score ranging from -1 to +1. Values near +1 indicate strongly positive language, values near -1 indicate strongly negative language, and values near 0 indicate neutral tone. VADER scores are unitless because they quantify textual sentiment rather than a physical measurement. Standard VADER thresholds classify scores below -0.05 as negative, above +0.05 as positive, and values in between as neutral. Since sentiment is continuous, it can be used directly in regression, correlation, and time-series analyses.

The categorical variable sentiment_cat converts these scores into positive, negative, or neutral groups for visualization and hypothesis testing. The date variable records article publication date in YYYY-MM-DD format and is useful for merging with the layoffs dataset, allowing direct comparison between sentiment trends and layoff activity over time. The boolean variable is_layoff_news identifies articles specifically related to layoffs, allowing filtration of irrelevant technology news prior to analysis.

The major limitations of this dataset are the relatively small sample size across the 2020-present time span and sparse temporal coverage after merging with the layoffs dataset. In addition, VADER is a lexicon-based model that measures surface-level lexical tone rather than deeper contextual meaning, sarcasm, or nuanced financial discussion. Aggregating sentiment by date may also reduce variability because multiple articles with different tones become compressed into a single daily sentiment measure.

### Dataset #3: Federal Funds Rate (FRED, 2020–Present)

This dataset contains 862 monthly observations of the effective Federal Funds Rate sourced directly from the Federal Reserve Bank of St. Louis (FRED), covering July 1954 through April 2026. For our analysis we use only the 76 observations from January 2020 onward, which align with the time span of the layoff events dataset.

The dataset has two columns:
- `observation_date`: First day of each month in YYYY-MM-DD format. Used as the join key after converting to a year-month period.
- `FEDFUNDS`: The monthly average effective Federal Funds Rate expressed as a percentage (e.g., `5.33` means 5.33%). This is our primary macroeconomic predictor variable.

Over the 2020–2026 window, the rate ranged from a near-zero floor of 0.05% during the COVID-era stimulus period (April 2020) to a peak of 5.33% in mid-2023 as the Fed aggressively tightened monetary policy to combat inflation. The mean rate across this period is 2.80% with a standard deviation of 2.18%, reflecting the sharp tightening cycle beginning in March 2022.

There are no missing values. Because the data is monthly and the layoff events are daily, we will merge on a year-month key: each layoff event inherits the FFR value for the month in which it occurred.

**Shortcomings:**
- Monthly granularity means all events within the same month share the same FFR value, which may smooth over intra-month variation.
- The FFR is a lagging policy instrument; its effect on corporate hiring decisions may operate with a delay of several months, which a simple contemporaneous merge does not capture.

```python
import pandas as pd
import matplotlib.pyplot as plt

# A. Load raw FFR data
ffr = pd.read_csv('data/00-raw/fedfunds.csv', parse_dates=['observation_date'])

# B. Filter to 2020-present to match layoff dataset
ffr = ffr[ffr['observation_date'] >= '2020-01-01'].copy()
ffr = ffr.rename(columns={'observation_date': 'date', 'FEDFUNDS': 'ffr'})

# C. Create year-month join key
ffr['year_month'] = ffr['date'].dt.to_period('M')

print(f"Shape: {ffr.shape[0]} rows × {ffr.shape[1]} columns")
print(f"Date range: {ffr['date'].min().date()} to {ffr['date'].max().date()}")
print(f"Missing values: {ffr.isnull().sum().sum()}")
print(f"\nFFR summary:")
print(ffr['ffr'].describe().round(2))

# D. Save to interim
ffr.to_csv('data/01-interim/fedfunds_clean.csv', index=False)
print("\nSaved: data/01-interim/fedfunds_clean.csv")

# E. Plot FFR over time
plt.figure(figsize=(11, 3))
plt.plot(ffr['date'], ffr['ffr'], color='#2c7bb6', linewidth=2)
plt.axvline(pd.Timestamp('2022-03-01'), color='gray', linestyle='--', alpha=0.7, label='Rate hikes begin (Mar 2022)')
plt.fill_between(ffr['date'], ffr['ffr'], alpha=0.15, color='#2c7bb6')
plt.xlabel('Date')
plt.ylabel('Federal Funds Rate (%)')
plt.title('Federal Funds Rate (2020–2026)')
plt.legend()
plt.tight_layout()
plt.show()

# F. Merge FFR into layoffs dataset on year-month
df_layoffs = pd.read_csv('data/01-interim/layoffs_events_clean.csv', parse_dates=['date'])
df_layoffs['year_month'] = df_layoffs['date'].dt.to_period('M')
df_with_ffr = df_layoffs.merge(ffr[['year_month', 'ffr']], on='year_month', how='left')

print(f"\nLayoffs with FFR attached: {len(df_with_ffr)} rows")
print(f"Missing FFR after merge: {df_with_ffr['ffr'].isnull().sum()}")

# G. Save processed dataset
df_with_ffr['log_layoff_count'] = __import__('numpy').log1p(df_with_ffr['layoff_count'])
df_with_ffr['sector'] = df_with_ffr['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})
df_with_ffr.to_csv('data/02-processed/layoffs_with_ffr.csv', index=False)
print("Saved: data/02-processed/layoffs_with_ffr.csv")
```

#### Layoff Scale Outliers

As mentioned above, our distribution is strongly right-skewed and contains some large outliers. However, these outliers are real reported events and thus should be kept. We will find outliers on raw `layoff_count` using the interquartile rule (values above Q3 + 1.5×IQR and Q3 + 3×IQR). We will also compare counts by sector. We then use `log(1 + layoff_count)` in modeling to reduce skew influence.


```python
df_out = pd.read_csv('data/02-processed/layoffs_events_clean.csv', parse_dates=['date'])
df_out['sector'] = df_out['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})
df_out['log_layoff_count'] = np.log1p(df_out['layoff_count'])

lc = df_out['layoff_count']
q1, q3 = lc.quantile(0.25), lc.quantile(0.75)
iqr = q3 - q1
fence_mild = q3 + 1.5 * iqr
fence_extreme = q3 + 3 * iqr

mild_mask = lc > fence_mild
extreme_mask = lc > fence_extreme

print('layoff_count outlier summary')
print(f'  N events: {len(df_out):,}')
print(f'  Q1 = {q1:.0f}, Q3 = {q3:.0f}, IQR = {iqr:.0f}')
print(f'  Mild outlier threshold (Q3 + 1.5×IQR): {fence_mild:.0f}  →  {mild_mask.sum()} events ({100*mild_mask.mean():.1f}%)')
print(f'  Extreme outlier threshold (Q3 + 3×IQR):   {fence_extreme:.0f}  →  {extreme_mask.sum()} events ({100*extreme_mask.mean():.1f}%)')
print()
print('Top 8 extreme layoff events:')
display_cols = ['company', 'date', 'layoff_count', 'sector']
print(df_out.loc[extreme_mask].nlargest(8, 'layoff_count')[display_cols].to_string(index=False))
print()
print('Extreme outliers by sector:')
print(df_out.loc[extreme_mask, 'sector'].value_counts().to_string())

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

bp_data = [
    df_out.loc[df_out['sector'] == 'Traditional Tech', 'log_layoff_count'],
    df_out.loc[df_out['sector'] == 'AI Company', 'log_layoff_count'],
]
axes[0].boxplot(bp_data, vert=True, labels=['Traditional Tech', 'AI Company'], patch_artist=True)
axes[0].set_ylabel('log(1 + layoff_count)')
axes[0].set_title('Layoff Scale by Sector (log scale)')
axes[0].grid(axis='y', alpha=0.3)

colors = np.where(extreme_mask, '#c0392b', '#5b8db8')
axes[1].scatter(df_out['date'], df_out['layoff_count'], c=colors, alpha=0.5, s=18, edgecolors='none')
axes[1].set_yscale('log')
axes[1].set_xlabel('Date')
axes[1].set_ylabel('layoff_count (log y-axis)')
axes[1].set_title('Layoff Events Over Time (red = Q3 + 3×IQR)')
axes[1].grid(alpha=0.3)

from matplotlib.lines import Line2D
legend_elems = [
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#5b8db8', markersize=8, label='Typical event'),
    Line2D([0], [0], marker='o', color='w', markerfacecolor='#c0392b', markersize=8, label='Extreme outlier'),
]
axes[1].legend(handles=legend_elems, loc='upper left')

plt.suptitle('Section 1 — Outliers in Layoff Count', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

Extreme events appear in both tech and AI sectors. However, it's worth noting that AI layoff events that are extreme outliers account for a bigger proportion of the AI layoff data compared to tech, which may suggest more common large restructuring rounds at AI companies. #### Section 2: Predictor Variables vs. Layoff Scale — Motivating the Regression

Before fitting a joint model, we examine each predictor's bivariate relationship with `log_layoff_count` by sector. This reveals whether the raw correlations hold across groups and motivates why multiple regression is needed to separate their independent effects.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Load the FFR-merged layoffs dataset
df = pd.read_csv('data/02-processed/layoffs_with_ffr.csv', parse_dates=['date'])

# Build full merged dataset (layoffs + FFR + sentiment)
news = pd.read_csv('data/01-interim/news_sentiment_clean.csv', parse_dates=['date'])
daily_sentiment = news.groupby('date')['sentiment'].mean().reset_index()
df_merged = pd.merge(df, daily_sentiment, on='date', how='inner')

# Save for reuse
df_merged.to_csv('data/02-processed/layoffs_full_merged.csv', index=False)

print(f'FFR dataset:       {len(df)} rows')
print(f'Full merged (FFR + sentiment): {len(df_merged)} rows')
print(f"  AI Company: {(df_merged['is_ai_company']==True).sum()} | "
      f"Traditional Tech: {(df_merged['is_ai_company']==False).sum()}")
```

```python
# ── Figure 4: FFR vs. log_layoff_count by sector ────────────────────────────
from scipy import stats
import numpy as np

colors = {'AI Company': '#e07b54', 'Traditional Tech': '#5b8db8'}

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
for ax, sector in zip(axes, ['Traditional Tech', 'AI Company']):
    subset = df[df['sector'] == sector]
    ax.scatter(subset['ffr'], subset['log_layoff_count'],
               alpha=0.35, s=18, color=colors[sector])
    slope, intercept, r, p, _ = stats.linregress(subset['ffr'], subset['log_layoff_count'])
    x_range = np.linspace(subset['ffr'].min(), subset['ffr'].max(), 100)
    ax.plot(x_range, slope * x_range + intercept, color='black', linewidth=1.8)
    ax.set_title(f'{sector}\nr={r:.3f}, p={p:.3f}, n={len(subset)}')
    ax.set_xlabel('Federal Funds Rate (%)')
    if sector == 'Traditional Tech':
        ax.set_ylabel('log(1 + Employees Laid Off)')

plt.suptitle('Figure 4: Federal Funds Rate vs. Log Layoff Count by Sector', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# ── Figure 6: Sentiment vs. log_layoff_count by sector ──────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)
for ax, sector in zip(axes, ['Traditional Tech', 'AI Company']):
    subset = df_merged[df_merged['sector'] == sector].dropna(
        subset=['sentiment', 'log_layoff_count'])
    ax.scatter(subset['sentiment'], subset['log_layoff_count'],
               alpha=0.5, s=25, color=colors[sector])
    ax.axvline(0, color='gray', linestyle='--', alpha=0.5, linewidth=0.8)
    if len(subset) > 2:
        slope, intercept, r, p, _ = stats.linregress(
            subset['sentiment'], subset['log_layoff_count'])
        x_range = np.linspace(subset['sentiment'].min(), subset['sentiment'].max(), 100)
        ax.plot(x_range, slope * x_range + intercept, color='black', linewidth=1.8)
        ax.set_title(f'{sector}\nr={r:.3f}, p={p:.3f}, n={len(subset)}')
    else:
        ax.set_title(f'{sector}\n(n={len(subset)}, insufficient)')
    ax.set_xlabel('VADER Sentiment Score')
    if sector == 'Traditional Tech':
        ax.set_ylabel('log(1 + Employees Laid Off)')

plt.suptitle('Figure 5: News Sentiment vs. Log Layoff Count by Sector', fontsize=13)
plt.tight_layout()
plt.show()
```

#### Predictors vs. Outcome Outliers

Before interpreting the bivariate scatters (Figures 4–5), we check whether `ffr`, `sentiment`, or `log_layoff_count` contain unusual values and outliers on the 73-row inner-joined sample used for sentiment analysis.
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_s2 = pd.read_csv('data/02-processed/layoffs_full_merged.csv', parse_dates=['date'])
df_s2['sector'] = df_s2['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})
df_s2['log_layoff_count'] = np.log1p(df_s2['layoff_count'])

def iqr_outlier_mask(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return (series < lo) | (series > hi), lo, hi

print(f'Inner-joined sample: n = {len(df_s2)}')
print(f"  Traditional Tech: {(df_s2['sector']=='Traditional Tech').sum()} | AI Company: {(df_s2['sector']=='AI Company').sum()}")
print()

for col in ['ffr', 'sentiment', 'log_layoff_count']:
    mask, lo, hi = iqr_outlier_mask(df_s2[col])
    print(f'{col}: IQR fence [{lo:.3f}, {hi:.3f}] → {mask.sum()} outliers')
    if mask.any():
        print(df_s2.loc[mask, ['company', 'date', col, 'sector']].sort_values(col).to_string(index=False))
    print()

oob = (df_s2['sentiment'] < -1) | (df_s2['sentiment'] > 1)
print(f'Sentiment outside [-1, 1]: {oob.sum()}')

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, col, title in zip(
    axes,
    ['ffr', 'sentiment', 'log_layoff_count'],
    ['Federal Funds Rate (%)', 'Daily mean sentiment', 'log(1 + layoff_count)'],
):
    ax.boxplot(
        [df_s2.loc[df_s2['sector'] == s, col] for s in ['Traditional Tech', 'AI Company']],
        labels=['Traditional Tech', 'AI Company'],
        patch_artist=True,
    )
    ax.set_title(title)
    ax.grid(axis='y', alpha=0.3)

plt.suptitle('Section 2 — Outliers on Merged Analysis Sample (n=73)', fontsize=13, y=1.05)
plt.tight_layout()
plt.show()
```

No sentiment values fall outside VADER's valid [-1, 1] range, which is expected. Our log_layoff_count shows 6 outliers, but the ones in the AI sector are likely not actually outliers since there are so few AI event data in this small 73-event sample (only 9 of the 73 rows are AI).

#### Section 3: Multiple Regression — Predicting Layoff Scale

The bivariate plots in Section 2 examined each predictor in isolation. Here we fit two OLS regression models using scikit-learn to assess the *joint* contribution of FFR and sentiment while controlling for company type:

- **Model 1 (main effects):** `log_layoff_count ~ ffr + sentiment + is_ai_company`
- **Model 2 (with interactions):** adds `ffr × is_ai_company` and `sentiment × is_ai_company`
- **Model 3 (with company controls):** adds raised_amount and stage (funding round) as controls to assess if company-specific financial maturity accounts for the differences in layoff scale.

The interaction terms are the direct test of our hypothesis: a significant `sentiment × is_ai_company` coefficient would indicate that AI companies respond differently to sentiment shifts than traditional tech companies do. We compute p-values manually from residuals since scikit-learn does not provide them natively.

> **Note:** Both models use the 73-row inner-joined dataset (layoffs matched to news dates). With only 9 AI company observations, interaction term estimates carry high uncertainty and should be treated as exploratory.

```python
import pandas as pd
import numpy as np
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# ── Load data ────────────────────────────────────────────────────────────────
df = pd.read_csv('data/02-processed/layoffs_full_merged.csv').dropna(
    subset=['log_layoff_count', 'ffr', 'sentiment']
)
df['is_ai'] = df['is_ai_company'].astype(int)
df['ffr_x_ai']  = df['ffr']       * df['is_ai']
df['sent_x_ai'] = df['sentiment'] * df['is_ai']
y = df['log_layoff_count'].values

# ── Helper: compute p-values from sklearn model ──────────────────────────────
def regression_summary(model, X, y, feature_names):
    n, k = X.shape
    y_pred = model.predict(X)
    residuals = y - y_pred
    RSS = np.sum(residuals ** 2)
    s2  = RSS / (n - k - 1)
    X_aug = np.column_stack([np.ones(n), X])
    var_b = s2 * np.linalg.inv(X_aug.T @ X_aug).diagonal()
    se = np.sqrt(var_b)
    coefs = np.append(model.intercept_, model.coef_)
    t_vals = coefs / se
    p_vals = 2 * (1 - stats.t.cdf(np.abs(t_vals), df=n - k - 1))
    names = ['intercept'] + feature_names
    summary = pd.DataFrame({
        'coef': coefs, 'std_err': se, 't': t_vals, 'p_value': p_vals
    }, index=names).round(4)
    summary['sig'] = summary['p_value'].apply(
        lambda p: '***' if p < 0.001 else ('**' if p < 0.01 else ('*' if p < 0.05 else ''))
    )
    return summary, r2_score(y, y_pred)

# ── Model 1: main effects ────────────────────────────────────────────────────
feat1 = ['ffr', 'sentiment', 'is_ai']
X1 = df[feat1].values
m1 = LinearRegression().fit(X1, y)
s1, r2_1 = regression_summary(m1, X1, y, feat1)

print('=' * 55)
print(f'Model 1 — Main Effects   R² = {r2_1:.4f}  (n={len(y)})')
print('=' * 55)
print(s1.to_string())

# ── Model 2: with interactions ───────────────────────────────────────────────
feat2 = ['ffr', 'sentiment', 'is_ai', 'ffr_x_ai', 'sent_x_ai']
X2 = df[feat2].values
m2 = LinearRegression().fit(X2, y)
s2, r2_2 = regression_summary(m2, X2, y, feat2)

print()
print('=' * 55)
print(f'Model 2 — With Interactions   R² = {r2_2:.4f}  (n={len(y)})')
print('=' * 55)
print(s2.to_string())

print(f'\nΔR² (Model 2 − Model 1) = {r2_2 - r2_1:.4f}')

# ── Model 3: with company controls ───────────────────────────────────────────

# 1. Create a subset of data that ONLY has the rows where we have all the info
# This prevents the "NaN" error
cols_to_check = ['ffr', 'sentiment', 'is_ai', 'raised_mm', 'stage']
df_clean_m3 = df.dropna(subset=cols_to_check)

# 2. Now create the dummies using the cleaned dataframe
stage_dummies = pd.get_dummies(df_clean_m3['stage'], prefix='stage', drop_first=True)
core_features = df_clean_m3[['ffr', 'sentiment', 'is_ai', 'raised_mm']]

X3 = pd.concat([core_features, stage_dummies], axis=1).astype(float)
y3 = df_clean_m3['log_layoff_count'] # Make sure the 'y' matches the 'X' length

# 3. Fit the model (using X3 and y3)
feat3_names = list(X3.columns)
m3 = LinearRegression().fit(X3, y3)
s3, r2_3 = regression_summary(m3, X3, y3, feat3_names)

print("-" * 55)
print(f"Model 3 - With Company Controls  R^2 = {r2_3:.4f} (n={len(y)})")
print("-" * 55)
print(s3.to_string())

# Calculate R-squared improvement over Model 1
print(f"\ndelta R^2 (Model 3 - Model 1): {r2_3 - r2_1:.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

df_s3 = pd.read_csv('data/02-processed/layoffs_full_merged.csv', parse_dates=['date'])
df_s3 = df_s3.dropna(subset=['layoff_count', 'ffr', 'sentiment']).copy()
df_s3['log_layoff_count'] = np.log1p(df_s3['layoff_count'])
df_s3['is_ai'] = df_s3['is_ai_company'].astype(int)
df_s3['sector'] = df_s3['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})

X = df_s3[['ffr', 'sentiment', 'is_ai']].values
y = df_s3['log_layoff_count'].values
model = LinearRegression().fit(X, y)
y_hat = model.predict(X)
resid = y - y_hat
std_resid = resid / resid.std()

df_s3['fitted'] = y_hat
df_s3['residual'] = resid
df_s3['std_residual'] = std_resid

influential = np.abs(std_resid) > 2
print(f'Model 1 residual outliers (|standardized residual| > 2): {influential.sum()} of {len(df_s3)}')
if influential.any():
    cols = ['company', 'date', 'sector', 'layoff_count', 'ffr', 'sentiment', 'fitted', 'std_residual']
    print(df_s3.loc[influential, cols].sort_values('std_residual').to_string(index=False, float_format=lambda x: f'{x:.3f}'))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].scatter(y_hat, resid, c=np.where(influential, '#c0392b', '#5b8db8'), alpha=0.7, edgecolors='white', linewidths=0.3)
axes[0].axhline(0, color='gray', linestyle='--')
axes[0].axhline(2 * resid.std(), color='gray', linestyle=':', alpha=0.7)
axes[0].axhline(-2 * resid.std(), color='gray', linestyle=':', alpha=0.7)
axes[0].set_xlabel('Fitted log(1 + layoff_count)')
axes[0].set_ylabel('Residual')
axes[0].set_title('Residuals vs. Fitted (Model 1)')
axes[0].grid(alpha=0.3)

axes[1].hist(std_resid, bins=15, color='steelblue', edgecolor='white')
axes[1].axvline(-2, color='#c0392b', linestyle='--', label='±2 SD')
axes[1].axvline(2, color='#c0392b', linestyle='--')
axes[1].set_xlabel('Standardized residual')
axes[1].set_ylabel('Count')
axes[1].set_title('Distribution of Standardized Residuals')
axes[1].legend()
axes[1].grid(axis='y', alpha=0.3)

plt.suptitle('Section 3 — Regression Outlier Diagnostics', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

We keep all rows in the reported models but treat coefficient p-values as exploratory.

**Findings — Multiple Regression**

**Model 1 (Main Effects, R² = 0.157):**
After controlling for all three predictors simultaneously, `ffr` loses all significance (coef = -0.048, p = 0.884). This reveals that the positive FFR–layoff correlation seen in Section 2 was a **time confound**: high interest rates and large layoffs co-occurred in 2022–2023, but FFR has no independent predictive power once sentiment and company type are held constant. `sentiment` remains significant (coef = -1.935, p = 0.034): more negative news coverage is independently associated with larger layoff events across both sectors. `is_ai` is also significant (coef = +1.630, p = 0.007): AI companies produce larger layoff events on average, consistent with the distribution findings in Section 1.

**Model 2 (With Interactions, R² = 0.219, ΔR² = +0.062):**
Adding interaction terms improves model fit modestly. The critical results for hypothesis testing are:
- `ffr_x_ai` (coef = +3.888, p = 0.027\*): **Significant.** AI companies respond far more strongly to FFR changes than traditional tech. The effective FFR slope for AI companies is -0.173 + 3.888 = **+3.715**, meaning a 1-percentage-point rise in the federal funds rate is associated with substantially larger layoff events in the AI sector, while traditional tech shows no significant FFR response. - `sent_x_ai` (coef = -3.769, p = 0.293): **Not significant.** The hypothesis predicted that AI companies would show a stronger sensitivity to news sentiment than traditional tech. While the direction is consistent (more negative), the differential effect cannot be statistically distinguished from zero given the available data.

**Model 3 (With Company Controls, $R^2 = 0.561$, $\Delta R^2 = +0.404$):**
Adding financial controls provides a more nuanced view of the layoff drivers.
- Company Maturity: The stage dummy variables indicate whether specific funding rounds (e.g., Series B vs. Late Stage) are more resilient or susceptible to market shifts.
- Capital Buffer: The raised_mm coefficient tests whether a larger "cash runway" mitigates the scale of layoffs. If significant, it suggests that layoffs are driven by liquidity constraints rather than just macro sentiment.
- Stability of Interaction Terms: The inclusion of these controls helps determine if the ffr_x_ai interaction remains significant. If the coefficient for ffr_x_ai holds steady, it strengthens the argument that AI companies have a unique structural sensitivity to interest rates that isn't just a byproduct of their funding stage.

**Hypothesis Evaluation:**
The results do not support the original hypothesis. Rather than AI companies being more sentiment-driven and traditional tech more macro-driven, the regression suggests the opposite pattern for FFR: AI companies exhibit a significantly stronger association with interest rate changes (ffr_x_ai p = 0.027), while the predicted sentiment advantage for AI companies remains undetectable (sent_x_ai p = 0.293). This may reflect that AI companies require large capital expenditures for infrastructure, making them more — not less — sensitive to the cost of borrowing.

**Limitations:**
These findings are exploratory and should be interpreted cautiously. The merged dataset contains only 73 observations, with just 9 AI company rows. The interaction terms are estimated with high standard errors, and the study is likely underpowered to detect a true sentiment differential if one exists. A larger matched dataset would be needed to draw definitive conclusions about the hypothesis.

### Analysis 1: Full-Sample Regression — Does FFR Predict Layoff Scale Differently by Sector?

Our hypothesis predicts that traditional tech layoffs are more sensitive to macroeconomic signals (Federal Funds Rate) than AI-focused companies. Since our EDA regression on the 73-row sentiment-matched sample is underpowered (only 9 AI events), we will instead fit OLS on the full FFR-matched dataset (around 1,590 events) using continuous FFR rather than a high/low split.

Our approach for this analysis is to do regression on the entire dataset with an interaction term.

Our models will be the following:
- **Model 1:** `log_layoff_count ~ ffr + is_ai_company`
- **Model 2:** `log_layoff_count ~ ffr + is_ai_company + ffr × is_ai_company`

As mentioned previously, our layoff scale is right-skewed and thus we use `log(1 + layoff_count)`. A significant `ffr × is_ai` coefficient would indicate the FFR slope differs between sectors which is a direct test of differential macro sensitivity.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# ── Load full FFR-matched sample ─────────────────────────────────────────────
df = pd.read_csv('data/02-processed/layoffs_with_ffr.csv', parse_dates=['date'])
df = df.dropna(subset=['ffr', 'layoff_count']).copy()
df['log_layoff_count'] = np.log1p(df['layoff_count'])
df['sector'] = df['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})
df['is_ai'] = df['is_ai_company'].astype(int)
df['ffr_x_ai'] = df['ffr'] * df['is_ai']
y = df['log_layoff_count'].values


def regression_summary(model, X, y, feature_names):
    n, k = X.shape
    y_pred = model.predict(X)
    residuals = y - y_pred
    RSS = np.sum(residuals ** 2)
    s2 = RSS / (n - k - 1)
    X_aug = np.column_stack([np.ones(n), X])
    se = np.sqrt(s2 * np.linalg.inv(X_aug.T @ X_aug).diagonal())
    coefs = np.append(model.intercept_, model.coef_)
    t_vals = coefs / se
    p_vals = 2 * (1 - stats.t.cdf(np.abs(t_vals), df=n - k - 1))
    names = ['intercept'] + feature_names
    summary = pd.DataFrame({
        'coef': coefs, 'std_err': se, 't': t_vals, 'p_value': p_vals
    }, index=names).round(4)
    summary['sig'] = summary['p_value'].apply(
        lambda p: '***' if p < 0.001 else ('**' if p < 0.01 else ('*' if p < 0.05 else ''))
    )
    return summary, r2_score(y, y_pred)


# ── Fit models ───────────────────────────────────────────────────────────────
feat1 = ['ffr', 'is_ai']
X1 = df[feat1].values
m1 = LinearRegression().fit(X1, y)
s1, r2_1 = regression_summary(m1, X1, y, feat1)

feat2 = ['ffr', 'is_ai', 'ffr_x_ai']
X2 = df[feat2].values
m2 = LinearRegression().fit(X2, y)
s2, r2_2 = regression_summary(m2, X2, y, feat2)

print(f'Full-sample regression (n = {len(df)})')
print('=' * 55)
print(f'Model 1 — Main effects   R² = {r2_1:.4f}')
print(s1.to_string())
print()
print(f'Model 2 — With interaction   R² = {r2_2:.4f}   ΔR² = {r2_2 - r2_1:.4f}')
print(s2.to_string())

# Sector-specific FFR slopes from Model 2
b_ffr = s2.loc['ffr', 'coef']
b_int = s2.loc['ffr_x_ai', 'coef']
print(f'\nEffective FFR slope — Traditional Tech: {b_ffr:.4f}')
print(f'Effective FFR slope — AI Company:       {b_ffr + b_int:.4f}')

# ── Figure: observed means + model-implied lines by sector ───────────────────
colors = {'AI Company': '#e07b54', 'Traditional Tech': '#5b8db8'}
fig, ax = plt.subplots(figsize=(8, 5))

for sector, is_ai in [('Traditional Tech', 0), ('AI Company', 1)]:
    sub = df[df['is_ai'] == is_ai]
    ax.scatter(sub['ffr'], sub['log_layoff_count'], alpha=0.15, s=12,
               color=colors[sector], label=f'{sector} (events)')
    ffr_grid = np.linspace(sub['ffr'].min(), sub['ffr'].max(), 50)
    pred = m2.intercept_ + m2.coef_[0] * ffr_grid + m2.coef_[1] * is_ai + m2.coef_[2] * ffr_grid * is_ai
    ax.plot(ffr_grid, pred, color=colors[sector], linewidth=2.5, label=f'{sector} (Model 2 fit)')

ax.set_xlabel('Federal Funds Rate (%)')
ax.set_ylabel('log(1 + layoff_count)')
ax.set_title('Analysis 1: FFR vs. Layoff Scale by Sector (Full Sample)')
ax.legend(loc='upper left', fontsize=9)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

**Findings — Full-Sample FFR Regression**

For our 1,590 FFR-matched layoff events, we can see from the positive slope that higher Federal Funds Rates are associated with larger layoffs overall (Model 1: `ffr` coef ≈ +0.09, p < 0.001). From our data, it seems that company type alone does not significantly shift layoff scale once FFR is included (Model 1: `is_ai` p ≈ 0.52).

Adding the interaction term (Model 2) raises R² only slightly (ΔR² ≈ 0.002) but points toward **steeper FFR sensitivity for AI companies**: the effective FFR slope is ~0.08 for traditional tech versus ~0.18 for AI companies. The interaction coefficient is marginally significant (p ≈ 0.08), so we treat this as suggestive rather than definitive but it is actually **opposite** to what we predicted in our hypothesis where we thought traditional tech would be more macro-sensitive.

Together with Analysis 2 (where AI companies also show a larger Cohen's d in high-FFR periods), the macro channel appears at least as strong for AI firms as for traditional tech. This could possibly be because expensive AI infrastructure makes high-rate environments especially costly and thus more macro-sensitive.


### Analysis 2: Do AI and Traditional Tech Companies Respond Differently to High vs Low Interest Rate Environments?

Our hypothesis predicts that traditional tech companies are more sensitive to macroeconomic conditions — specifically the Federal Funds Rate — than AI-focused companies. To test this, we split all layoff events into **High FFR** and **Low FFR** periods using the median FFR as the cutoff, then ask: within each sector, does the scale of layoffs differ significantly between high- and low-rate environments?

We run an **independent-samples t-test** on `log_layoff_count` for each sector separately:

- **H₀:** mean log_layoff_count in High FFR period = mean log_layoff_count in Low FFR period (for a given sector)  
- **H₁:** mean log_layoff_count in High FFR period ≠ mean log_layoff_count in Low FFR period

We use **Cohen's d** as the effect size to compare the *magnitude* of the FFR effect across sectors. The hypothesis predicts a larger Cohen's d for Traditional Tech than for AI companies.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# ── Load data ────────────────────────────────────────────────────────────────
df = pd.read_csv('data/02-processed/layoffs_with_ffr.csv', parse_dates=['date'])
df['sector'] = df['is_ai_company'].map({True: 'AI Company', False: 'Traditional Tech'})
df['log_layoff_count'] = np.log1p(df['layoff_count'])
df = df.dropna(subset=['ffr', 'layoff_count'])

# ── Split into High / Low FFR using the median as cutoff ─────────────────────
ffr_median = df['ffr'].median()
df['ffr_period'] = np.where(df['ffr'] >= ffr_median, 'High FFR', 'Low FFR')
print(f'FFR median cutoff: {ffr_median:.2f}%')
print(df.groupby(['sector', 'ffr_period']).size().unstack(), '\n')

# ── Cohen's d helper ─────────────────────────────────────────────────────────
def cohen_d(a, b):
    pooled_std = np.sqrt((a.std(ddof=1)**2 + b.std(ddof=1)**2) / 2)
    return (a.mean() - b.mean()) / pooled_std

# ── Run t-tests and compute effect sizes ─────────────────────────────────────
results = []
for sector in ['AI Company', 'Traditional Tech']:
    sub  = df[df['sector'] == sector]
    high = sub[sub['ffr_period'] == 'High FFR']['log_layoff_count']
    low  = sub[sub['ffr_period'] == 'Low FFR']['log_layoff_count']
    t, p = stats.ttest_ind(high, low)
    d    = cohen_d(high, low)
    results.append({'sector': sector, 'n_high': len(high), 'n_low': len(low),
                    'mean_high': high.mean(), 'mean_low': low.mean(),
                    't': t, 'p': p, 'cohens_d': d})
    print(f'{sector}')
    print(f'  High FFR: mean={high.mean():.3f}  n={len(high)}')
    print(f'  Low  FFR: mean={low.mean():.3f}  n={len(low)}')
    print(f'  t={t:.3f}, p={p:.4f}, Cohen\'s d={d:.3f}')
    print()

# ── Figure: grouped violin plot ───────────────────────────────────────────────
colors = {'AI Company': '#e07b54', 'Traditional Tech': '#5b8db8'}
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

for ax, r in zip(axes, results):
    sector = r['sector']
    sub    = df[df['sector'] == sector]
    groups = [sub[sub['ffr_period'] == p]['log_layoff_count'] for p in ['Low FFR', 'High FFR']]
    vp = ax.violinplot(groups, positions=[0, 1], showmedians=True)
    for body in vp['bodies']:
        body.set_facecolor(colors[sector])
        body.set_alpha(0.6)
    ax.set_xticks([0, 1])
    ax.set_xticklabels([f'Low FFR\n(< {ffr_median:.2f}%)', f'High FFR\n(≥ {ffr_median:.2f}%)'])
    ax.set_title(f'{sector}\nt={r["t"]:.3f}, p={r["p"]:.4f}, d={r["cohens_d"]:.3f}', fontsize=11)
    ax.set_ylabel('log(layoff_count + 1)' if ax == axes[0] else '')
    ax.set_facecolor('#f9f9f9')

fig.suptitle('Layoff Scale in High vs Low FFR Periods by Sector', fontsize=13, y=1.01)
fig.tight_layout()
plt.show()
```

**Findings — FFR Sensitivity by Sector**

Using the median FFR of 4.09% as the cutoff, both sectors show statistically significant differences in layoff scale between high and low interest rate environments:

| Sector | Mean log-layoff (Low FFR) | Mean log-layoff (High FFR) | t | p | Cohen's d |
|--------|--------------------------|---------------------------|---|---|----------|
| AI Company | 4.285 | 4.805 | 2.479 | 0.014 | 0.342 |
| Traditional Tech | 4.459 | 4.748 | 4.334 | <0.001 | 0.234 |

Both sectors experience significantly larger layoff events during high-FFR periods, consistent with macroeconomic pressure driving cost-cutting across the board. However, the hypothesis predicted that Traditional Tech would be more FFR-sensitive — the data shows the opposite: AI companies have a larger effect size (d = 0.342) compared to Traditional Tech (d = 0.234).

This null result (no differential sensitivity in the predicted direction) likely reflects the nature of the 2022–2023 rate hike cycle, which coincided with a broad industry-wide correction affecting both sectors simultaneously. AI companies, many of which were in growth-stage funding rounds, may have been especially exposed to tightening capital markets during high-rate periods, amplifying their layoff response beyond what the hypothesis anticipated.

## Discussion and Conclusion

In this project, we analyzed the dynamics of the tech sector layoffs between 2020 and 2026. Specifically, we contrasted layoffs in AI-focused companies with those in traditional big tech companies to infer the macroeconomic impacts of changes in the Federal Funds Rate (FFR) and media sentiment. Our primary analysis revealed that layoffs are significantly right-skewed and that traditional tech companies experienced more frequent layoffs and restructurings than AI firms. Moreover, our independent-samples t-test confirmed that both traditional tech and AI companies experienced more layoffs in higher-FFR environments than in lower-FFR environments. This confirms that macroeconomic changes are influential in company restructuring.

Building on our prior research, in 2020, there was massive growth in tech companies, and a lot of funding was available due to low interest rates. In 2022, however, the Federal Reserve raised interest rates to fight inflation, causing many tech companies to lose their funding and conduct massive, highly publicized layoffs. While most companies were suffering, AI companies still had investors and massive corporations investing in AI technology, simultaneously acting as a cushion for many workers in this industry.

Although FFR had a clear correlation with tech layoffs, we also assumed that media sentiment would have a strong influence. We hypothesized that if the news about a company's economy was strongly negative, it would directly predict more frequent layoffs. However, our regression models showed no statistically significant predictive power of news sentiment. We believe this is likely due to the fact that VADER (a tool to measure the positivity or negativity of news and media) was not catching early signs, but the hindsight commentary of reporters writing negative articles after the terrible incidents had already occurred. Another reason is that our data didn’t include many days where a layoff event and clear news sentiment data overlapped, so the model couldn’t identify the news articles that were impacted by economic forces like the rise of FFR. Ultimately, the interest rates had a higher impact than the news articles.

There were several key limitations that must be accounted for when interpreting these results. First, our data was very small and had a small sample size of n=73. Out of those, 64 were traditional tech companies, and only 9 were AI companies. As a result, the results must be treated with caution because there is a lack of statistical power. Second, this data is an analysis of a small period of time, mostly situated during layoff events. This is not longitudinal data, so we cannot conclude the aftermath of these layoff events. Finally, every company from our list was forced into two categories: AI Company or Traditional Tech. This may seem oversimplified in cases such as Microsoft, Google, and Meta, which are both traditional tech giants and also large investors in the AI industry.

If this investigation were to continue outside the scope of this class, one of the largest priorities would be to find a better way to label each company rather than just using the two simplified categories. One way to do this would be to analyze how much a company invests in AI and what is a significant amount of money for that company to be more AI-focused. It is also necessary to gain more extensive data. Using web scraping to pull in more news from all over the world would solve the problem of data loss when trying to inner-join layoff dates and news articles data. Finally, it would be more statistically significant and impactful to look at the macroeconomic and media sentiment effects over a longer period of time. When interest rates rise, companies don’t usually lay off workers the next day. It may take a couple of months for the impacts to start showing. Ultimately, our research is limited to a narrow scope; collecting more data over longer periods of time would provide a clearer picture of exactly what is happening in the layoff events at tech companies.
